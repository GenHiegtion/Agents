### __Generator__

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import AIMessage, HumanMessage
from langchain_openai import ChatOpenAI

generation_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "Bạn là một thi sĩ giỏi chuyên viết thơ lục bát Tiếng Việt. Nhiệm vụ của bạn là sáng tác những bài thơ lục bát xuất sắc."
            " Hãy tạo ra bài thơ lục bát hay nhất tuân thủ CHÍNH XÁC luật thơ: câu lục (6 chữ) và câu bát (8 chữ) xen kẽ, vần điệu đúng quy tắc (câu chẵn vần với nhau)."
            " Nếu người dùng đưa ra lời phê bình về luật thơ hoặc nội dung, hãy phản hồi bằng phiên bản đã sửa lỗi hoàn toàn.",
        ),
        MessagesPlaceholder(variable_name="messages"),
    ]
)

llm = ChatOpenAI(model="gpt-4o")
generate_chain = generation_prompt | llm
```

```python
request = HumanMessage(
    content="Hãy viết một bài thơ lục bát 4 câu (2 câu lục, 2 câu bát) về trí tuệ nhân tạo và tương lai của loài người."
)
response = generate_chain.invoke({"messages": [request]})
response.content
```

### __Reflector__

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

reflection_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "Bạn là một giáo sư văn học chuyên về thơ ca Việt Nam, đặc biệt là thể thơ lục bát. Hãy đưa ra lời phê bình NGHIÊM KHẮC và chi tiết về bài thơ lục bát."
            " Kiểm tra kỹ: (1) Số chữ mỗi câu (6-8-6-8), (2) Vần điệu (câu 2 vần với câu 4), (3) Thanh điệu, (4) Ý nghĩa và hình ảnh thơ, (5) Tính mạch lạc."
            " Chỉ ra CHÍNH XÁC từng lỗi sai và yêu cầu sửa lại. Không bỏ qua bất kỳ lỗi nào về luật thơ.",
        ),
        MessagesPlaceholder(variable_name="messages"),
    ]
)

reflect_chain = reflection_prompt | llm
```

```python
reflected_response = reflect_chain.invoke({"messages": [request, HumanMessage(content=response.content)]})
reflected_response.content
```

### Repeat

```python
reflected_response = reflect_chain.invoke({"messages": [request, AIMessage(content=response.content), HumanMessage(content=reflected_response.content)]})
reflected_response.content
```

### __Graph__

```python
from typing import List, Sequence
from langchain_core.messages import BaseMessage
from langgraph.graph import END, MessageGraph

REFLECT = "reflect"
GENERATE = "generate"


def generation_node(state: Sequence[BaseMessage]):
    return generate_chain.invoke({"messages": state})


def reflection_node(messages: Sequence[BaseMessage]) -> List[BaseMessage]:
    res = reflect_chain.invoke({"messages": messages})
    return [HumanMessage(content=res.content)]


builder = MessageGraph()
builder.add_node(GENERATE, generation_node)
builder.add_node(REFLECT, reflection_node)
builder.set_entry_point(GENERATE)


def should_continue(state: List[BaseMessage]):
    if len(state) > 6:
        return END
    return REFLECT


builder.add_conditional_edges("generate", should_continue, {END:END, "reflect":"reflect"})
builder.add_edge(REFLECT, GENERATE)
graph = builder.compile()
```

```python
request = HumanMessage(
    content="Hãy viết một bài thơ lục bát 4 câu (2 câu lục, 2 câu bát) về trí tuệ nhân tạo và tương lai của loài người."
)
response = graph.invoke(request)
response
```
